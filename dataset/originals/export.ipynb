```python
import pandas as pd
import numpy as np
```

```python
df_sales = pd.read_csv(filepath_or_buffer="./salesDB_grocery_market/sales.csv", 
                       delimiter=";",
                       low_memory=False)
```

```python
del df_sales["TransactionNumber"]
del df_sales["TotalPrice"]
del df_sales["Discount"]
```

```python
df_products = pd.read_csv(filepath_or_buffer="./salesDB_grocery_market/products.csv",
                          delimiter=";",
                          low_memory=False)
```

```python
del df_products["ModifyDate"]
del df_products["Resistant"]
del df_products["IsAllergic"]
del df_products["VitalityDays"]
del df_products["Price"]
del df_products["Class"]
```

```python
df_category = pd.read_csv(filepath_or_buffer="./salesDB_grocery_market/categories.csv",
                          delimiter=";",
                          low_memory=False)
```

```python
df_products = df_products.merge(right=df_category, left_on="CategoryID", right_on="CategoryID", how="left")
```

```python
df_sales = df_sales.merge(right=df_products, left_on="ProductID", right_on="ProductID", how="left")
```

```python
del df_sales["SalesID"]
del df_sales["SalesPersonID"]
del df_sales["SalesDate"]
del df_sales["CategoryName"]
del df_sales["Quantity"]
```

```python
del df_sales["ProductName"]
del df_sales["ProductID"]
```

```python
# Normaliza para que o ID da categoria comece por 0
df_sales["CategoryID"] = df_sales["CategoryID"] - 1
```

```python
df_sales = df_sales.drop_duplicates()
df_sales = df_sales.sort_values(by="CustomerID")
df_sales = df_sales.reset_index(drop=True)
```

```python
df_sales = df_sales.sample(frac=1)
df_sales = df_sales[:1000000].copy()
```

```python
df_sales1 = df_sales.sample(frac=1)[:25000].copy()
df_sales2 = df_sales.sample(frac=1)[:50000].copy()
df_sales3 = df_sales.sample(frac=1)[:100000].copy()
```

```python
df_sales1 = df_sales1.sort_values(by="CustomerID")
df_sales2 = df_sales2.sort_values(by="CustomerID")
df_sales3 = df_sales3.sort_values(by="CustomerID")
```

```python
df_sales1.to_csv("sales1.csv", index=False)
df_sales2.to_csv("sales2.csv", index=False)
df_sales3.to_csv("sales3.csv", index=False)
```
